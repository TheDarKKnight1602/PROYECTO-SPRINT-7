```python
import pandas as pd
import plotly.express as px
```

```python

df = pd.read_csv('../vehicles_us.csv')
```

```python
df.head()
```

```python
import plotly.io as pio
pio.renderers.default = "browser"  # Forzar que se abra en el navegador
```

```python
fig = px.histogram(df, x="odometer") # crear un histograma
fig.show()
```

```python
df['price'].plot.hist(bins=50, figsize=(10, 5), title='Distribución de Precios')
```

```python
df.plot.scatter(x='model_year', y='price', figsize=(10, 5), title='Precio vs. Año del Modelo')
```

```python
df['type'].value_counts().plot.bar(figsize=(10, 5), title='Cantidad de Vehículos por Tipo')
```

```python
df['condition'].value_counts().plot.pie(autopct='%1.1f%%', figsize=(6, 6), title='Proporción por Condición')
```
